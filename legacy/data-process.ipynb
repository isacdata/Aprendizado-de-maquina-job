- Random Forest
- Regressão Logística

## Random Forest

Necessário:
- Sem escalonamento
- Encoding One-hot ou Ordinal
- Tratamento de Outliers leve
- Suposição linear: Não

```python
random_forest_data = spi_indicators.copy()
random_forest_data = random_forest_data[random_forest_data['income'] != 'Not classified']
random_forest_data.describe()
```

```python
random_forest_data.columns
```

```python
X = random_forest_data.drop(columns=['income', 'iso3c', 'country'])
y = random_forest_data['income']
```

```python
random_forest_data.dtypes
```

```python
# Definindo quem é numérico e quem é categórico pelo que sobrou
numeric_features = [
    'year', 'population', 'overall_score', 'data_use_score', 
    'data_services_score', 'data_products_score', 
    'data_sources_score', 'data_infrastructure_score'
]
categorical_features = ['region']
```

```python
import matplotlib.pyplot as plt
```

```python
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer

numeric_transformer_rf = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median')) # Usa a mediana para não ser afetado por outliers da população
])

categorical_transformer_rf = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OneHotEncoder(handle_unknown='ignore'))
])

preprocessor_rf = ColumnTransformer(
    transformers=[
        ('num', numeric_transformer_rf, numeric_features),
        ('cat', categorical_transformer_rf, categorical_features)
    ])

# 4. Crie o Pipeline final (Exemplo com SVM)
clf_rf = Pipeline(steps=[('preprocessor', preprocessor_rf)])
```

## SVM e Regressão Logística

Necessário:
- Escalonamento
- Encoding One-hot ou Ordinal
- Tratamento de Outliers Pesado
- Suposição linear: Sim

```python

numeric_transformer_linear = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler()) # <-- CRÍTICO AQUI
])

categorical_transformer_linear = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
    # drop='first' remove a primeira coluna gerada para evitar multicolinearidade (ótimo para LogReg)
    ('encoder', OneHotEncoder(handle_unknown='ignore', drop='first')) 
])

preprocessor_linear = ColumnTransformer(
    transformers=[
        ('num', numeric_transformer_linear, numeric_features),
        ('cat', categorical_transformer_linear, categorical_features)
    ])
```

```python
clf_svm_log = Pipeline(steps=[('preprocessor', preprocessor_linear)])
```
